# training and testing process

```python
# after cleaning i have null missing values 
# so i decided to train the model
```

```python
from sklearn.model_selection import train_test_split
```

```python
X=df.iloc[:,:-1]
X
```

```python
y=df.iloc[:,-1]
y
```

```python
# now dividin the training and test set
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
```

```python
X_train
```

# Feature scaling

```python
# 1 simple feature scaling
# 2 standardization(z-score)
# 3 Normalization(min-max)
```

```python
from sklearn.preprocessing import StandardScaler

# 1. Convert columns 4 to 7 to float in the dataframe first
cols_to_scale = X_train.columns[4:7]
X_train[cols_to_scale] = X_train[cols_to_scale].astype('float64')

# 2. Apply StandardScaler and assign back cleanly
sc = StandardScaler()
X_train[cols_to_scale] = sc.fit_transform(X_train[cols_to_scale])

# 1. Convert columns 4 to 7 to float in X_test
cols_to_scale = X_test.columns[4:7]
X_test[cols_to_scale] = X_test[cols_to_scale].astype('float64')

# 2. Transform X_test using the already fitted scaler
X_test[cols_to_scale] = sc.transform(X_test[cols_to_scale])
```

# Selection of ML algorithm based on accuracy


```python
# applying the Logistic Regression 
from sklearn.linear_model import LogisticRegression
lr_model=LogisticRegression(random_state=42)
lr_model.fit(X_train,y_train)
print("train accuracy", lr_model.score(X_train,y_train))
print('test accuracy',lr_model.score(X_test,y_test))
```

```python
# now applying the RandomForest classifier
from sklearn.ensemble import RandomForestClassifier
rf_model=RandomForestClassifier(n_estimators=100,random_state=42)
rf_model.fit(X_train,y_train)
print('train accuracy',rf_model.score(X_train,y_train))
print('test accuracy',rf_model.score(X_test,y_test))
```

```python
# now applying support vector machine(SVM)
from sklearn.svm import SVC
svm_model=SVC(kernel='rbf',random_state=42)
svm_model.fit(X_train,y_train)
print('Train accuracy',svm_model.score(X_train,y_train))
print('Test accuracy',svm_model.score(X_test,y_test))
```

```python
# now apply KNN 
from sklearn.neighbors import KNeighborsClassifier
knn_model=KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train,y_train)
print('Train accuracy',knn_model.score(X_train,y_train))
print('Test accuracy',knn_model.score(X_test,y_test))
```

```python
# corss validation check
from sklearn.model_selection import cross_val_score
cv_scores=cross_val_score(rf_model,X_train,y_train,cv=5,scoring='accuracy')
print('Cross-Validation scores for each fold:',cv_scores)
print('Mean Cross-Validation Accuracy:{:.4f}'.format(cv_scores.mean()))
print('Standard deviation: {:.4f}'.format(cv_scores.std()))
```

```python
# now after applying 4 ml algorithms i can conclude that the best for my model is RandomForestClassifier because  i got 72 % accuray on test and on training i got 100%  
```
